### 1.2 Exploración de nulos

### 1.3 Depuración de atributos constantes

Algunas variables contienen el 100% el mismo valor, representando una constante. Estas son eliminadas ya que no aportan ninguna información.

```python
conteos = pd.DataFrame()
for i in variables:
    conteo = df[i].value_counts()
    porcentaje = df[i].value_counts(normalize=True)

    describir = pd.concat([conteo, porcentaje], axis = 1, keys =['Conteo','Porcentaje']).reset_index()
    describir['Variable'] = i 
    conteos = conteos.append(describir)
```

```python
print(conteos[conteos['Porcentaje']>=0.95].shape)
baja_varianza = list(conteos[conteos['Porcentaje']>=0.95]['Variable'])
conteos[conteos['Porcentaje']>=0.95].sort_values(by='Porcentaje',ascending=True).head(20)
```

```python
variables = [x for x in variables if x not in baja_varianza]
len(variables)
```

```python
df_depurado['porc_hab'] = np.where(df_depurado['r305']!=0, df_depurado['r306']/df_depurado['r305'],np.nan)
calculo_attrs1 = df_depurado.groupby('r004').agg({'r021a': ['mean', 'median', 'std'],
                                 'aproba1': ['mean', 'median', 'std'],
                                 'r305': ['mean', 'median', 'std'],
                                 'r306': ['mean', 'median', 'std'],
                                 'porc_hab': ['mean', 'median', 'std'],
                                 'money': ['mean', 'median', 'std'],
                                 'ingre': ['mean', 'median', 'std'],
                                 'ingpe': ['mean', 'median', 'std'],
                                 'ingfa': ['mean', 'median', 'std'],
                                 'gmem': ['mean', 'median', 'std'],
                                 'r328total': ['mean', 'median', 'std'],
                                 'gmvi': ['mean', 'median', 'std'],
                                 'imeds': ['mean', 'median', 'std']}).reset_index()
calculo_attrs1.columns = ['DEPTO','miembros_ph_avg','miembros_ph_med', 'miembros_ph_std',
                          'grados_avg','grados_med','grados_std',
                          'cnt_habit_ph_avg','cnt_habit_ph_med','cnt_habit_ph_std',
                          'cnt_dorm_ph_avg','cnt_dorm_ph_med','cnt_dorm_ph_std',
                          'porc_dorm_avg','porc_dorm_med','porc_dorm_std',
                          'ing_dep_avg','ing_dep_med','ing_dep_std',
                          'ing_emp_avg','ing_emp_med','ing_emp_std',
                          'ing_pc_avg','ing_pc_med','ing_pc_std',
                          'ing_fam_avg','ing_fam_med','ing_fam_std',
                          'gasto_emp_avg','gasto_emp_med','gasto_emp_std',
                          'gasto_serv_avg','gasto_serv_med','gasto_serv_std',
                          'gasto_viv_avg','gasto_viv_med','gasto_viv_std',
                          'ing_dep_avg','ing_dep_med','ing_dep_std']
calculo_attrs1
```

```python
# Normalizar valores
df_depurado['hogares_lectura'] = np.where(df_depurado['r202a'].isin(['Sí', 'Sí, sólo leer']),1,0)
df_depurado['hogares_pobres'] = np.where(df_depurado['pobreza'].isin(['Pobreza extrema', 'Pobreza relativa']),1,0)
df_depurado['hogares_techo_lamina'] = np.where(df_depurado['r302'].isin(['Lámina metálica']),1,0)
df_depurado['hogares_pared_concreto'] = np.where(df_depurado['r303'].isin(['Concreto o mixto']),1,0)
df_depurado['hogares_ladrillo'] = np.where(df_depurado['r304'].isin(['Ladrillo cerámico']),1,0)
df_depurado['hogares_propietarios'] = np.where(df_depurado['r308'].isin(['Propietario(a)','Propietario(a) de la vivienda pero la está pagando a plazos' ]),1,0)
df_depurado['hogares_inquilinos'] = np.where(df_depurado['r308'].isin(['Inquilino(a)' ]),1,0)
df_depurado['hogares_sin_agua'] = np.where(df_depurado['r312'].isin(['No tiene' ]),1,0)
df_depurado['hogares_gas'] = np.where(df_depurado['r320'].isin(['Gas propano' ]),1,0)
df_depurado['hogares_sin_seguro'] = np.where(df_depurado['r501'].isin(['No tiene' ]),1,0)
df_depurado['hogares_rurales'] = np.where(df_depurado['area'].isin(['Rural' ]),1,0)

crear = ['hogares_lectura','hogares_pobres','hogares_techo_lamina','hogares_pared_concreto','hogares_ladrillo','hogares_propietarios','hogares_inquilinos','hogares_sin_agua','hogares_gas','hogares_sin_seguro','hogares_rurales']
```

## Clusterizacion

```python
variables_cluster = [x for x in atributos_depto.columns if x not in ['DEPTO','CONTEO']]
len(variables_cluster)
```

```python
variables_cluster
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.cluster import KMeans, AgglomerativeClustering
from sklearn.metrics import silhouette_score
import matplotlib.pyplot as plt
```

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(atributos_depto[variables_cluster])
```

```python
pd.set_option('display.float_format', lambda x: '%.2f' % x)
```

```python
pca = PCA(n_components=4)
X_pca = pca.fit_transform(X_scaled)
for i in pca.explained_variance_ratio_:
    print(i)
```

```python
k_values = range(2, 11)     # k desde 2 hasta 10 (ajustable)
sil_scores = []

for k in k_values:
    model = AgglomerativeClustering(
        n_clusters=k,
        linkage="ward"          # ward funciona bien con datos continuos
    )
    labels = model.fit_predict(X_pca)
    sil = silhouette_score(X_pca, labels)
    sil_scores.append(sil)
    print(f"k={k} → Silhouette Score = {sil:.4f}")
```

```python
model_final = AgglomerativeClustering(
    n_clusters=3,
    linkage="ward"
)

atributos_depto["cluster"] = model_final.fit_predict(X_pca)
```

```python
pd.crosstab(index = atributos_depto.DEPTO, columns = atributos_depto.cluster)
```

```python
atributos_depto
```

### Exploración por cluster

```python
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import classification_report, confusion_matrix
from sklearn.tree import plot_tree

import matplotlib.pyplot as plt
```

```python
variables_cluster.remove('hogares_rurales')
```

```python
X = atributos_depto[variables_cluster]
y = atributos_depto['cluster']

# Separar en entrenamiento y prueba
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42, stratify=y
)


# ------------------------------------------------------------
# 2. Crear y entrenar el árbol de clasificación
# ------------------------------------------------------------
tree_model = DecisionTreeClassifier(
    criterion="gini",     # o "entropy"
    max_depth=5,          # controla complejidad
    random_state=42
)

tree_model.fit(X, y)
```

```python
y_pred = tree_model.predict(X_train)

print("=== CLASSIFICATION REPORT ===")
print(classification_report(y_train, y_pred))

print("=== CONFUSION MATRIX ===")
print(confusion_matrix(y_train, y_pred))
```

```python
y_pred = tree_model.predict(X_test)

print("=== CLASSIFICATION REPORT ===")
print(classification_report(y_test, y_pred))

print("=== CONFUSION MATRIX ===")
print(confusion_matrix(y_test, y_pred))
```

```python
plt.figure(figsize=(20, 10))   # ajusta el tamaño según lo grande del árbol

plot_tree(
    tree_model,
    feature_names=X.columns,
    class_names=[str(c) for c in np.unique(y)],
    filled=True,
    rounded=True,
    fontsize=10
)
plt.title("Árbol de Clasificación")
plt.show()
```

## Version 2 de clusterizacion


Hacemos la clusterizacion por persona y vemos qué correlacion tienen los clusters con los departamentos.

```python
atributos_cluster = ['porc_hab','hogares_rurales','hogares_sin_seguro','hogares_gas','hogares_sin_agua','hogares_inquilinos','hogares_propietarios','hogares_ladrillo','hogares_pared_concreto','hogares_techo_lamina','hogares_pobres',
                     'hogares_lectura','imeds','gmvi','r328total','gmem','ingfa','ingpe','ingre','money','r306','r305','aproba1','r021a']
```

```python
df_depurado[atributos_cluster] = df_depurado[atributos_cluster].fillna(0)
scaler = StandardScaler()
X_scaled = scaler.fit_transform(df_depurado[atributos_cluster])


pca = PCA(n_components=12)
X_pca = pca.fit_transform(X_scaled)
for i in pca.explained_variance_ratio_:
    print(i)
```

```python

k_values = range(2, 11)     # k desde 2 hasta 10 (ajustable)
sil_scores = []

for k in k_values:
    kmeans = KMeans(n_clusters=k, random_state=42)
    labels = kmeans.fit_predict(X_pca)
    sil = silhouette_score(X_pca, labels)
    sil_scores.append(sil)
    print(f"k={k} → Silhouette Score = {sil:.4f}")

```

```python

model_final = KMeans(
    n_clusters=3
)

df_depurado["cluster"] = model_final.fit_predict(X_pca)


pd.crosstab(index = df_depurado.r004, columns = df_depurado.cluster, normalize = 'columns')

```
