### 1.2 Exploración de nulos

Debido a la alta cantidad de variables y a que se sabe de antemano que muchas de ellas están poco pobladas; se hará primero una depuración preliminar para usar solamente aquellas variables que tengan un nivel mínimo de poblamiento.

```python
cols_nas = pd.DataFrame(df.isna().mean()).reset_index()
cols_nas.columns = ['Variable', 'nulos']

sns.histplot(data = cols_nas, x = 'nulos')
```

```python
#### Como se observa en la gráfica, hay varios atributos con 100% de nulos. Para el análisis, se tomarán aquellos que tengan < 50% de nulos.
print(cols_nas[cols_nas['nulos']<0.5].shape)

variables = list(cols_nas[cols_nas['nulos']<0.5]['Variable'])
print(len(variables))
cols_nas[cols_nas['nulos']<0.5].sort_values(by='nulos',ascending=False).head(20)
```

### 1.3 Depuración de atributos constantes

Algunas variables contienen el 100% el mismo valor, representando una constante. Estas son eliminadas ya que no aportan ninguna información.

```python
conteos = pd.DataFrame()
for i in variables:
    conteo = df[i].value_counts()
    porcentaje = df[i].value_counts(normalize=True)

    describir = pd.concat([conteo, porcentaje], axis = 1, keys =['Conteo','Porcentaje']).reset_index()
    describir['Variable'] = i 
    conteos = conteos.append(describir)
```

```python
print(conteos[conteos['Porcentaje']>=0.95].shape)
baja_varianza = list(conteos[conteos['Porcentaje']>=0.95]['Variable'])
conteos[conteos['Porcentaje']>=0.95].sort_values(by='Porcentaje',ascending=True).head(20)
```

```python
variables = [x for x in variables if x not in baja_varianza]
len(variables)
```

### 1.4 Depuración de registros

Se eliminan registros con muy poca información en las variables seleccionadas (hogares que no respondieron) o respondieron muy pocas preguntas

```python
df[variables].shape
```

```python
## Deben tener al menos un 80% de las preguntas contestadas
df_depurado = df[variables].dropna(thresh=df[variables].shape[1]*0.8,axis=0)
df_depurado.shape
```

```python
remover = ['lote','folio','idboleta']
variables = [x for x in variables if x not in remover]
len(variables)
```

# Atributos a nivel departamento

- Promedio de miembros por hogar (r021a - avg)
- promedio de grados aprobados aproba1
- promedio de numero de habitaciones por depto. r305
- promedio/mediana de # de dormitorios r306.
- avg median de 306/305
- avg median del valor en dinero
- promedio mediana de money, ingre, ingpe, ingfa, gme, r328total , gmvi, imeds



- porcentaje de hogares que sabe leer y escribir (r202a)
- porcentaje de hogares en probreza 
- porcentaje de hogares con techo de lamina (r302)
- porcentaje de hogares con paredes de concreto (r303)
- porcentaje de hogares con ladrillo cerámico (r304)
- porcentaje de viviendas ocupadas por propietarios (r308)
- porcentaje de viviendas ocupadas por inquilinos
- porcentaje de viviendas sin agua (r312)
- porcentaje de hogares con gas propano (r320)
- porcentaje de hogares con seguro medico (r501)


```python
df_depurado['porc_hab'] = np.where(df_depurado['r305']!=0, df_depurado['r306']/df_depurado['r305'],np.nan)
calculo_attrs1 = df_depurado.groupby('r004').agg({'r021a': ['mean', 'median'],
                                 'aproba1': ['mean', 'median'],
                                 'r305': ['mean', 'median'],
                                 'r306': ['mean', 'median'],
                                 'porc_hab': ['mean', 'median'],
                                 'money': ['mean', 'median'],
                                 'ingre': ['mean', 'median'],
                                 'ingpe': ['mean', 'median'],
                                 'ingfa': ['mean', 'median'],
                                 'gmem': ['mean', 'median'],
                                 'r328total': ['mean', 'median'],
                                 'gmvi': ['mean', 'median'],
                                 'imeds': ['mean', 'median']}).reset_index()
calculo_attrs1.columns = ['DEPTO','miembros_ph_avg','miembros_ph_med', 'grados_avg','grados_med',
                          'cnt_habit_ph_avg','cnt_habit_ph_med','cnt_dorm_ph_avg','cnt_dorm_ph_med',
                          'porc_dorm_avg','porc_dorm_med','ing_dep_avg','ing_dep_med',
                          'ing_emp_avg','ing_emp_med','ing_pc_avg','ing_pc_med','ing_fam_avg','ing_fam_med',
                          'gasto_emp_avg','gasto_emp_med','gasto_serv_avg','gasto_serv_med','gasto_viv_avg','gasto_viv_med',
                          'ing_dep_avg','ing_dep_med']
calculo_attrs1
```

```python
# Normalizar valores
df_depurado['hogares_lectura'] = np.where(df_depurado['r202a'].isin(['Sí', 'Sí, sólo leer']),1,0)
df_depurado['hogares_pobres'] = np.where(df_depurado['pobreza'].isin(['Pobreza extrema', 'Pobreza relativa']),1,0)
df_depurado['hogares_techo_lamina'] = np.where(df_depurado['r302'].isin(['Lámina metálica']),1,0)
df_depurado['hogares_pared_concreto'] = np.where(df_depurado['r303'].isin(['Concreto o mixto']),1,0)
df_depurado['hogares_ladrillo'] = np.where(df_depurado['r304'].isin(['Ladrillo cerámico']),1,0)
df_depurado['hogares_propietarios'] = np.where(df_depurado['r308'].isin(['Propietario(a)','Propietario(a) de la vivienda pero la está pagando a plazos' ]),1,0)
df_depurado['hogares_inquilinos'] = np.where(df_depurado['r308'].isin(['Inquilino(a)' ]),1,0)
df_depurado['hogares_sin_agua'] = np.where(df_depurado['r312'].isin(['No tiene' ]),1,0)
df_depurado['hogares_gas'] = np.where(df_depurado['r320'].isin(['Gas propano' ]),1,0)
df_depurado['hogares_sin_seguro'] = np.where(df_depurado['r501'].isin(['No tiene' ]),1,0)

crear = ['hogares_lectura','hogares_pobres','hogares_techo_lamina','hogares_pared_concreto','hogares_ladrillo','hogares_propietarios','hogares_inquilinos','hogares_sin_agua','hogares_gas','hogares_sin_seguro']
```

```python
pd.crosstab(index = df_depurado['hogares_lectura'], columns = df_depurado['r202a'])
```

```python
atributos_depto = df_depurado.groupby('r004').size().reset_index()
atributos_depto.columns = ['DEPTO','CONTEO']
atributos_depto
```

```python
for vari in crear:
    conteo = pd.crosstab(index = df_depurado['r004'], columns = df_depurado[vari], normalize = 'index').reset_index()
    conteo.columns = ['DEPTO','NO',vari]
    conteo = conteo[['DEPTO',vari]]
    atributos_depto = atributos_depto.merge(conteo, on = 'DEPTO')
    
```

```python
atributos_depto = atributos_depto.merge(calculo_attrs1, on = 'DEPTO')
```
